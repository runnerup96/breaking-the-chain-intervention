```python
import os
import json
from averitec_dataset import AVeriTeCDataset

path = '/home/chaichuk/frontdoor_llm_causality/intervention_analysis/intervention_predictions/averitec'
files = os.listdir(path)

dataset = AVeriTeCDataset('/home/chaichuk/frontdoor_llm_causality/statics/datasets/AVeriTeC/data')

for file in files:
    data = None
    with open(f'{path}/{file}', 'r', encoding='utf-8') as f:
        data = json.load(f)

    if 'result' not in data:
        continue

    generated = {}

    for i in range(len(data['result'])):
        if data['result'][i]['completion_type'] == 'gold_structure':
            continue
        idx = data['result'][i]['idx']
        supporting_questions = data['result'][i]['supporting_questions']
        label = data['result'][i]["label"]

        generated[idx] = {'supporting_questions': supporting_questions, 'label': label}

    
    new_data = {}

    for i in range(len(dataset)):
        golden = dataset[i]
        idx = golden['idx']
        if idx not in generated:
            continue
        gen = generated[idx]
        if gen['supporting_questions'] == {}:
            continue

        if gen['supporting_questions'] != golden['supporting_questions']:
            new_data[idx] = {'claim': golden['claim'],
                             'explanations': golden['explanations'],
                             'golden_supporting_questions': golden['supporting_questions'],
                             'golden_label': golden['label'],
                             'bad_supporting_questions': gen['supporting_questions'],
                             'bad_label': gen['label']}
            
    print(file, len(new_data))
            
    with open(f'/home/chaichuk/frontdoor_llm_causality/statics/datasets/averitec_correction/{file}', 'w', encoding='utf-8') as f:
        json.dump(new_data, f, indent=4, ensure_ascii=False)
            
    
```

```python
import os
import json
from ricechem_dataset import RiceChemDataset

path = '/home/chaichuk/frontdoor_llm_causality/intervention_analysis/intervention_predictions/ricechem'
files = ['gemma2-2B.json']#os.listdir(path)

dataset = RiceChemDataset('/home/chaichuk/frontdoor_llm_causality/statics/datasets/RiceChem/data')

for file in files:
    data = None
    with open(f'{path}/{file}', 'r', encoding='utf-8') as f:
        data = json.load(f)

    if 'result' not in data:
        continue

    generated = {}

    for i in range(len(data['result'])):
        if data['result'][i]['completion_type'] == 'gold_structure':
            continue
        idx = data['result'][i]['idx']
        filled_rubric = data['result'][i]['filled_rubric']
        score = data['result'][i]['score']

        generated[idx] = {'filled_rubric': filled_rubric, 'score': score}

    
    new_data = {}

    for i in range(len(dataset)):
        golden = dataset[i]
        idx = golden['idx']
        if idx not in generated:
            continue
        gen = generated[idx]
        if gen['filled_rubric'] == {}:
            continue

        if set(gen['filled_rubric'].keys()) != set(golden['filled_rubric'].keys()):
            continue

        if gen['filled_rubric'] != golden['filled_rubric']:
            new_data[idx] = {'task_idx': golden['task_idx'], 
                             'task': golden['task'],
                             'score_range': golden['score_range'],
                             'student_answer': golden['student_answer'],
                             'golden_rubric': golden['filled_rubric'],
                             'golden_score': golden['score'],
                             'bad_rubric': gen['filled_rubric'],
                             'bad_score': gen['score']}
            
    print(file, len(new_data))
            
    with open(f'/home/chaichuk/frontdoor_llm_causality/statics/datasets/ricechem_correction/{file}', 'w', encoding='utf-8') as f:
        json.dump(new_data, f, indent=4, ensure_ascii=False)
            
    
```
